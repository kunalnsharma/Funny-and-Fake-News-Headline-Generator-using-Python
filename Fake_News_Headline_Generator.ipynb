```python
import random
```

```python
subjects=[
    'Shahrukh Khan',
    'Virat Kohli',
    'Nirmala Sitaraman',
    'A Mumbai Cat',
    'PM Modi',
    'A group of Monkeys',
    'Auto Rikshaw Driver from Delhi'
]

actions = [
    'launches',
    'cancels',
    'dances with',
    'eats',
    'declares war on',
    'orders',
    'celebrates'
]

places_or_things = [
    'at red fort',
    'mumbai local train',
    'inside parliament',
    'at Ganga Ghat',
    'during IPL Match',
    'India Gate',
    'Lal Kuan'
]

while True:
  subject = random.choice(subjects)
  action = random.choice(actions)
  place_or_thing = random.choice(places_or_things)

  headline = (f"BREAKING NEWS : {subject} {action} {place_or_thing}")
  print("\n" + headline)

  user_input = input("\n Do you want another Headline (Yes/No)").strip().lower()
  if user_input == 'no':
    break
```
